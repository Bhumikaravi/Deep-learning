Gradient Descent Implementation

Description

Students will implement a simple Gradient Descent algorithm to understand how a model updates its parameters and reduces prediction error during training.
Task 1: Initialize a Simple Dataset and Model Parameters
We use a simple dataset where the relationship between input and output is approximately:

y = 2x

```python
import numpy as np
import matplotlib.pyplot as plt

# Simple dataset
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8, 10], dtype=float)

# Initialize model parameters
m = 0.0       # slope
b = 0.0       # intercept

# Learning rate
learning_rate = 0.01

# Number of iterations
iterations = 100

print("Initial slope (m):", m)
print("Initial intercept (b):", b)
```

Task 2: Implement Gradient Descent
Code

```python
# Store loss values
loss_history = []

# Gradient Descent
for i in range(iterations):

    # Make predictions
    y_pred = m * X + b

    # Calculate error
    error = y_pred - y

    # Calculate Mean Squared Error
    loss = np.mean(error ** 2)

    # Store loss
    loss_history.append(loss)

    # Calculate gradients
    dm = (2 / len(X)) * np.sum(error * X)
    db = (2 / len(X)) * np.sum(error)

    # Update parameters
    m = m - learning_rate * dm
    b = b - learning_rate * db

print("Final slope (m):", m)
print("Final intercept (b):", b)
print("Final loss:", loss_history[-1])
```

Gradient Descent works in the following steps:
1. The model makes predictions.
2. The prediction error is calculated.
3. The loss is calculated.
4. Gradients are calculated.
5. Model parameters are updated.
6. The process is repeated for multiple iterations

Task 3: Display and Analyze Loss Reduction

```python
# Plot loss reduction
plt.figure(figsize=(8, 5))
plt.plot(range(1, iterations + 1), loss_history, marker='o')

plt.xlabel("Iterations")
plt.ylabel("Loss")
plt.title("Gradient Descent - Loss Reduction")
plt.grid(True)
plt.show()
```

Expected Graph

The graph should show a decreasing curve, because the loss becomes smaller as the model learns better parameter values.

```python
# Final predictions
y_pred = m * X + b

print("Actual values:   ", y)
print("Predicted values:", y_pred)
```

Analysis
- At the beginning, the loss is high because the parameters are not trained.
- As the iterations increase, the loss decreases.
- This shows that Gradient Descent is updating the parameters to improve predictions.
- After many iterations, the loss becomes very small, meaning the model's predictions are close to the actual values.

Result: The graph clearly shows the reduction in error during training, demonstrating how Gradient Descent helps the model learn.
